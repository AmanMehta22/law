```python
import sys
from pathlib import Path

RAG_ROOT = Path.cwd().parent

sys.path.append(str(RAG_ROOT))
```

```python
from src.vectorStore import VectorStoreManager
from src.retriever import RAGRetriever


# Load existing ChromaDB
vector_manager = VectorStoreManager()
vector_store = vector_manager.load()

print(
    "Documents in Chroma:",
    vector_store._collection.count()
)


# Create retriever
retriever = RAGRetriever(
    vector_store=vector_store,
    k=5
)


# Test query
query = "I bought a defective laptop and I want my money back"

results = retriever.retrieve(query)

print("\nResults:", len(results))

for i, doc in enumerate(results, start=1):

    print("\n" + "=" * 70)
    print(f"RESULT {i}")
    print("=" * 70)

    print("Concept ID:", doc.metadata.get("concept_id"))
    print("Type:", doc.metadata.get("concept_type"))
    print("Title:", doc.metadata.get("title"))
```
